```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "name": ["Riya", "Aman", "Neha", "Karan", "Isha", "Rohan"],
    "city": ["Pune", "Mumbai", "Pune", "Kolhapur", "Mumbai", "Pune"],
    "marks": [85, 62, 91, 45, 78, 70],
    "attendance": [92, 75, 98, 60, 88, 80]
})

print(df)

plt.bar(df["name"], df["marks"])
plt.title("Student Marks")
plt.xlabel("Name")
plt.ylabel("Marks")

plt.show()
```

```python
#groupby then plot
df.groupby("city")["marks"].mean().plot(kind="bar")
plt.title("Average Marks by City")

plt.ylabel("Average Marks")
plt.show()
```

```python
#create a figure with 2 rows and 2 columns of subplots showing 4 different charts
# Create a figure with 2 rows and 2 columns of subplots showing 4 different charts
import matplotlib.pyplot as plt
import numpy as np


x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
categories = ["A", "B", "C", "D"]
values = [10, 25, 15, 30]
scatter_x = np.random.rand(20)
scatter_y = np.random.rand(20)
hist_data = np.random.randn(100)


fig, axs = plt.subplots(2, 2, figsize=(10, 8))


axs[0, 0].plot(x, y, marker="o", color="blue")
axs[0, 0].set_title("Line Chart")
axs[0, 0].set_xlabel("X")
axs[0, 0].set_ylabel("Y")

axs[0, 1].bar(categories, values, color="orange")
axs[0, 1].set_title("Bar Chart")
axs[0, 1].set_xlabel("Category")
axs[0, 1].set_ylabel("Value")


axs[1, 0].scatter(scatter_x, scatter_y, color="green")
axs[1, 0].set_title("Scatter Plot")
axs[1, 0].set_xlabel("X")
axs[1, 0].set_ylabel("Y")


axs[1, 1].hist(hist_data, bins=10, color="purple", edgecolor="black")
axs[1, 1].set_title("Histogram")
axs[1, 1].set_xlabel("Value")
axs[1, 1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
#Draw a heatmap of the correlation matrix of a data frame with 3 numeric column and  save it as heatmap.png


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "marks": [85, 62, 91, 45, 78, 70],
    "attendance": [92, 75, 98, 60, 88, 80],
    "game_hours": [4, 6, 3, 5, 2, 4]
})
corr = df.corr()
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Matrix")

# Save the plot
plt.savefig("heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create mock data for 6 months of a shop
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 15000, 13500, 18000, 16500, 21000]
orders = [240, 310, 260, 380, 300, 420]

df_shop = pd.DataFrame({
    "Month": months,
    "Revenue": revenue,
    "Orders": orders
})

# Bonus: calculate average revenue per order
df_shop["Avg_Revenue_Per_Order"] = df_shop["Revenue"] / df_shop["Orders"]

# 2. Print the best month and the total revenue
total_revenue = df_shop["Revenue"].sum()
best_month_idx = df_shop["Revenue"].idxmax()
best_month = df_shop.loc[best_month_idx, "Month"]
best_month_rev = df_shop.loc[best_month_idx, "Revenue"]

print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Best Month: {best_month} (${best_month_rev:,.2f})")

# 3. Create Dashboard Figure
sns.set_theme(style="whitegrid")
fig, axs = plt.subplots(3, 1, figsize=(10, 12))

# Chart 1: Line Chart - Monthly Revenue
sns.lineplot(data=df_shop, x="Month", y="Revenue", marker="o", color="b", ax=axs[0], linewidth=2.5)
axs[0].set_title("Monthly Revenue (6 Months)", fontsize=14, fontweight="bold")
axs[0].set_xlabel("Month")
axs[0].set_ylabel("Revenue ($)")

# Chart 2: Bar Chart - Orders Per Month
sns.barplot(data=df_shop, x="Month", y="Orders", palette="viridis", ax=axs[1])
axs[1].set_title("Orders Per Month", fontsize=14, fontweight="bold")
axs[1].set_xlabel("Month")
axs[1].set_ylabel("Number of Orders")

# Chart 3: Bonus Chart - Average Revenue per Order
sns.barplot(data=df_shop, x="Month", y="Avg_Revenue_Per_Order", palette="magma", ax=axs[2])
axs[2].set_title("Average Revenue per Order", fontsize=14, fontweight="bold")
axs[2].set_xlabel("Month")
axs[2].set_ylabel("Avg Revenue / Order ($)")

plt.tight_layout()

# 4. Save the dashboard
plt.savefig("dashboard.png", dpi=150, bbox_inches="tight")
plt.show()
```
