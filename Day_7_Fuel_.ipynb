```python
import pandas as pd

df = pd.read_csv("student_data_1000.csv")

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Python_Marks"] = df["Python_Marks"].fillna(df["Python_Marks"].mean())
df["SQL_Marks"] = df["SQL_Marks"].fillna(df["SQL_Marks"].mean())
df["Math_Marks"] = df["Math_Marks"].fillna(df["Math_Marks"].mean())
df["Total_Marks"] = df["Total_Marks"].fillna(df["Total_Marks"].mean())
df["Attendance_Percent"] = df["Attendance_Percent"].fillna(df["Attendance_Percent"].mean())

print(df.isnull().sum())
```

```python
#outlier
print(df["Age"].median())
print(df["Age"].mean())
```

```python
q1 = df["Age"].quantile(0.25)
q3 = df["Age"].quantile(0.75)
iqr = q3 - q1
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr

print("Q1, Q3, IQR:", q1, q3, iqr)
print("Limits:", low, high)


print(df[(df["Age"] < low) | (df["Age"] > high)])
```

```python
#add duplicate row to a dataframe ,count them and remove them
import pandas as pd

# 1. Create a sample DataFrame
df = pd.DataFrame({
    "name": ["Riya", "Aman", "Karan"],
    "age": [21, 22, 23]
})

# 2. Add an explicit duplicate row (copying the first row)
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

print("--- DataFrame with Duplicate Row ---")
print(df)

# 3. Count duplicate rows using .duplicated().sum()
print("\nNumber of duplicate rows:", df.duplicated().sum())

# 4. Remove duplicates using .drop_duplicates()
df = df.drop_duplicates().reset_index(drop=True)

print("\n--- After Removing Duplicates ---")
print(df)
```
