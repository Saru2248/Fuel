```python
#Apply a 10% discount to prices above 500 in an array using np.where
import numpy as np

prices = np.array([200, 450, 600, 800, 1000])

discounted = np.where(prices > 500, prices * 0.90, prices)

print("Original prices:", prices)
print("Final prices:", discounted)
```

```python
#normalize an array using (x-min)/(max-min)
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print("Original array:", arr)
print("Normalized array:", normalized)
```

```python
#compare the time taken by a list and a numpy array to add 1,000,000 number
import numpy as np
import time

numbers_list = list(range(1000000))
numbers_array = np.arange(1000000)


start = time.time()
list_result = [x + 1 for x in numbers_list]
list_time = time.time() - start


start = time.time()
array_result = numbers_array + 1
array_time = time.time() - start

print(f"List time: {list_time:.6f} seconds")
print(f"NumPy array time: {array_time:.6f} seconds")

if array_time > 0:
    print(f"NumPy is {list_time / array_time:.2f} times faster")
```

```python
import numpy as np

# Take student names as input
names = []

for i in range(5):
    name = input(f"Enter name of student {i + 1}: ")
    names.append(name)

# Take marks for 3 subjects
marks = []

for i in range(5):
    print(f"\nEnter marks for {names[i]}:")
    s1 = float(input("Subject 1: "))
    s2 = float(input("Subject 2: "))
    s3 = float(input("Subject 3: "))

    marks.append([s1, s2, s3])

# Convert lists into NumPy arrays
names = np.array(names)
marks = np.array(marks)

# 1. Total and average of each student
total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)

print("\n--- Student Results ---")

for i in range(5):
    print(f"{names[i]}: Total = {total[i]}, Average = {average[i]:.2f}")

# 2. Find the topper
topper_index = np.argmax(total)

print(f"\nTopper: {names[topper_index]}")
print(f"Total Marks: {total[topper_index]}")

# 3. Average marks of each subject
subject_average = np.mean(marks, axis=0)

print("\n--- Subject Averages ---")
print(f"Subject 1: {subject_average[0]:.2f}")
print(f"Subject 2: {subject_average[1]:.2f}")
print(f"Subject 3: {subject_average[2]:.2f}")

# 4. Count students who passed all subjects
passed_all = np.all(marks >= 50, axis=1)
pass_count = np.sum(passed_all)

print(f"\nStudents who passed all subjects: {pass_count}")

# 5. Assign grades using np.where()
grades = np.where(
    average >= 90, "A",
    np.where(average >= 75, "B",
    np.where(average >= 50, "C", "F"))
)

print("\n--- Student Grades ---")

for i in range(5):
    print(f"{names[i]}: {grades[i]}")
```
