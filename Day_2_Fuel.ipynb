```python
marks = int(input("Enter marks: "))

if marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
elif marks >= 60:
    print("C")
elif marks >= 50:
    print("D")
else:
    print("Fail")
```

```python
age = int(input("Enter age: "))
has_id = input("Do you have ID? (yes/no): ")

if age >= 18 and has_id == "yes":
    print("Entry Allowed")

if age < 13 or age > 60:
    print("Discount")

if has_id == "no":
    print("ID required")
```
